```python
import os

# Path to your extracted image folder
base_path = "/content/drive/MyDrive/crop_project/Original_Image"

image_paths = []

for _, row in df.iterrows():

    image_name = row["Image Name"] + ".jpg"
    class_name = row["Class Name"]

    image_path = os.path.join(base_path, class_name, image_name)

    image_paths.append(image_path)

df["Image_Path"] = image_paths

print(df[["Image Name", "Image_Path"]].head())
```

```python
import cv2
import numpy as np

IMG_SIZE = 224

images = []

for path in df["Image_Path"]:

    img = cv2.imread(path)

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))

    img = img / 255.0

    images.append(img)

images = np.array(images, dtype=np.float32)

print(images.shape)
```

```python
from sklearn.model_selection import train_test_split

X_img_train, X_img_test, X_weather_train, X_weather_test, y_train, y_test = train_test_split(
    images,
    weather_features,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

print("Train Images :", X_img_train.shape)
print("Test Images  :", X_img_test.shape)

print("Train Weather:", X_weather_train.shape)
print("Test Weather :", X_weather_test.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, GlobalAveragePooling2D
from tensorflow.keras.applications import InceptionV3

# Image input
image_input = Input(shape=(224, 224, 3), name="Image_Input")

# InceptionV3 backbone
base_model = InceptionV3(
    weights="imagenet",
    include_top=False,
    input_tensor=image_input
)

# Freeze pretrained layers initially
base_model.trainable = False

# Extract image features
image_features = GlobalAveragePooling2D()(base_model.output)

image_features = Dense(256, activation="relu")(image_features)

print("Image branch created successfully!")
```

```python
from tensorflow.keras.layers import Input, Dense

weather_input = Input(shape=(2,), name="Weather_Input")

weather_features = Dense(32, activation="relu")(weather_input)
weather_features = Dense(16, activation="relu")(weather_features)

print("Weather branch created successfully!")
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

print("All libraries imported successfully!")
```

```python
df = pd.read_csv(
    "/content/drive/MyDrive/crop_project/metadata.csv",
    encoding="latin1"
)

print(df.head())
print(df.shape)
```

```python
df = df.dropna(subset=["Image Name"])
df = df.reset_index(drop=True)

print(df.shape)
```

```python
df["Temperature"] = (
    df["Temperature"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["Temperature"] = pd.to_numeric(df["Temperature"])

print(df["Temperature"].head())
```

```python
print(df.shape)
print(df["Temperature"].head())
```

```python
# Convert weather names to numbers
from sklearn.preprocessing import LabelEncoder

weather_encoder = LabelEncoder()
df["Weather_Encoded"] = weather_encoder.fit_transform(df["Weather"])

# Convert disease names to numbers
label_encoder = LabelEncoder()
df["Label"] = label_encoder.fit_transform(df["Class Name"])

# Check the encoded values
print(df[["Weather", "Weather_Encoded"]].drop_duplicates())
print()
print(df[["Class Name", "Label"]].drop_duplicates())
```

```python
# Import os library
import os

# Folder containing all class folders
base_path = "/content/drive/MyDrive/crop_project/Original_Image"

# Create image path list
image_paths = []

# Create full path for each image
for _, row in df.iterrows():

    image_name = row["Image Name"] + ".jpg"
    class_name = row["Class Name"]

    image_path = os.path.join(base_path, class_name, image_name)

    image_paths.append(image_path)

# Add image paths to dataframe
df["Image_Path"] = image_paths

# Show first 5 paths
print(df[["Image Name", "Image_Path"]].head())
```

```python
# Check missing images
missing = df[~df["Image_Path"].apply(os.path.exists)]

print("Missing Images:", len(missing))
```

```python
from tqdm import tqdm

# Image size
IMG_SIZE = 224

# Store all images
images = []

# Read images with progress bar
for path in tqdm(df["Image_Path"]):

    img = cv2.imread(path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
    img = img / 255.0

    images.append(img)

# Convert to NumPy array
images = np.array(images, dtype=np.float32)

print(images.shape)
```
